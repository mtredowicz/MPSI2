## Zadanie 10. Ten sam objaw, różne przyczyny
Regresja $\hat y=ax+b$ na $n=10$ punktach ($x=1,\dots,10$, $y\approx2x+1$ z szumem), GD, $\eta=0{,}01$, $200$ kroków, start $(0,0)$. Poniżej pięć wersji funkcji `grad(a, b)` (kod w komórce poniżej).

Dla każdej **przewidź** przebieg straty, zanim uruchomisz:

* **A**: poprawny gradient $\mathrm{MSE}$;
* **B**: gradient z pominiętym dzieleniem przez $n$ (suma zamiast średniej);
* **C**: gradient z odwróconym znakiem;
* **D**: gradient bez czynnika $2$;
* **E**: gradient po $a$ poprawny, a składowa po $b$ stale równa $0$.

a) Które objawy można pomylić? Którego błędu *nie widać* na wykresie straty?
b) Czy B i D to „błędy gradientu”? Gradient *czego* liczą?
c) Jak je odróżnić od poprawnego kodu przed uczeniem?

**Odpowiedź / notatki:**

```python
import numpy as np
x = np.arange(1, 11.); y = 2*x + 1 + np.array([0.5, -0.4, 0.3, -0.2, 0.6, -0.5, 0.1, -0.3, 0.4, -0.6]); n = len(x)
mse = lambda a, b: np.mean((a*x + b - y)**2)
def gA(a, b): r = a*x + b - y; return 2/n*(r*x).sum(), 2/n*r.sum()
def gB(a, b): r = a*x + b - y; return 2*(r*x).sum(), 2*r.sum()
def gC(a, b): r = a*x + b - y; return -2/n*(r*x).sum(), -2/n*r.sum()
def gD(a, b): r = a*x + b - y; return 1/n*(r*x).sum(), 1/n*r.sum()
def gE(a, b): r = a*x + b - y; return 2/n*(r*x).sum(), 0.0
def run(g, eta=0.01, N=200):
    a = b = 0.0; h = [mse(a, b)]
    for _ in range(N):
        ga, gb = g(a, b); a -= eta*ga; b -= eta*gb; h.append(mse(a, b))
        if not np.isfinite(h[-1]) or h[-1] > 1e100: break
    return a, b, h
```
