## Zadanie 8. Polowanie na błędy
Poniżej cztery operacje z błędem w `_wstecz`. Dla każdej: (i) znajdź błąd, (ii) podaj najprostszą funkcję i punkt, na których gradcheck go **wykryje**, (iii) podaj funkcję i punkt, na których gradcheck go **nie** wykryje, i wyjaśnij dlaczego.

```python
def mul_zly(x, y):
    w = Wartosc(x.dane * y.dane, (x, y))
    def _wstecz():
        x.grad += x.dane * w.grad        # A
        y.grad += y.dane * w.grad
    w._wstecz = _wstecz; return w

def pow_zly(x, p):                      # B
    w = Wartosc(x.dane ** p, (x,))
    def _wstecz():
        x.grad += p * x.dane ** p * w.grad
    w._wstecz = _wstecz; return w

def exp_zly(x):                         # C
    w = Wartosc(math.exp(x.dane), (x,))
    def _wstecz():
        x.grad += math.exp(x.dane)
    w._wstecz = _wstecz; return w

def relu_zly(x):                        # D
    w = Wartosc(max(0.0, x.dane), (x,))
    def _wstecz():
        x.grad += (1.0 if x.dane >= 0 else 0.0) * w.grad
    w._wstecz = _wstecz; return w
```
