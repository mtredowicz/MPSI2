## Zasady

* Co tydzień nowy notebook z zadaniami; w domu przygotowujesz je na następne zajęcia. LLM jest dozwolony.
* Przed zajęciami uzupełniasz **Excel**: zaznaczasz zadania, które zrobiłeś/aś.
* Na zajęciach prowadząca wyrywkowo wybiera zadanie i osobę, która je zadeklarowała, i omawiamy je w dyskusji (dlaczego tak, a co jeśli inaczej…).
* **Punkt za zadanie** dostaje ten, kto je zadeklarował i (jeśli został zapytany) potrafi je wyjaśnić i odpowiedzieć na pytania. Jeśli nie potrafi, nie ma punktu za to zadanie.
* Nieobecność = brak punktów za zadania z tych zajęć.
* Do oceny końcowej **nie liczą się 2 notebooki z najmniejszą liczbą punktów** (czyli 2 nieobecności są dozwolone).
* Punkty za notebooki = 60 · (zdobyte punkty za zadania) / (wszystkie zadania w liczonych notebookach).
* Kolokwia: pisemne, 45 min, bez komputera i LLM.
