## Zadanie 6. Cichy błąd
Poniższy kod liczy gradient funkcji $f(x,y)=xy+x^2$ w punkcie $(3,2)$ przebiegiem wstecz. Uruchom go, porównaj wynik z ilorazem różnicowym, znajdź błąd i popraw.

**Odpowiedź / notatki:**

````python
# graf: v1 = x*y ; v2 = x*x ; f = v1 + v2    w punkcie (x, y) = (3, 2)
x, y = 3.0, 2.0
v1 = x*y; v2 = x*x; f = v1 + v2

# przebieg wstecz
f_b = 1.0
v1_b = f_b; v2_b = f_b                  # dodawanie rozdziela
x_b = v1_b * y                          # przez v1
x_b = v2_b * x                          # przez v2 (pierwszy czynnik)
x_b = v2_b * x                          # przez v2 (drugi czynnik)
y_b = v1_b * x
print(f, x_b, y_b)
````
